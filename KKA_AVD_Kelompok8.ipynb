```python
import io
import pandas as pd

nama_file = 'DataTugasKelompok.csv'

# 1. Buka file menggunakan variabel nama_file
with open(nama_file, 'r', encoding='utf-8') as f:
  lines = [line.strip().strip('"') for line in f if line.strip()]

# 2. Load ke DataFrame
df = pd.read_csv(io.StringIO('\n'.join(lines)))

# 3. Hitung statistik deskriptif
total_karyawan = len(df)
rata_umur = df['Age'].mean()
median_umur = df['Age'].median()

rata_gaji = df['MonthlyIncome'].mean()
median_gaji = df['MonthlyIncome'].median()

dept_terbanyak = df['Department'].value_counts().idxmax()
dept_count = df['Department'].value_counts().max()
dept_persen = (dept_count / total_karyawan) * 100

attrition_count = (df['Attrition'] == 'Yes').sum()
attrition_rate = (attrition_count / total_karyawan) * 100

# 4. Cetak hasil
print(f'Total Karyawan         : {total_karyawan} orang')
print(
    f'Rata-rata Usia         : {rata_umur:.2f} tahun (Median:'
    f' {median_umur:.0f} tahun)'
)
print(f'Rata-rata Gaji Bulanan : ${rata_gaji:,.2f} (Median: ${median_gaji:,.2f})')
print(
    f'Departemen Dominan     : {dept_terbanyak} ({dept_count} orang /'
    f' {dept_persen:.1f}%)'
)
print(
    f'Tingkat Attrition      : {attrition_rate:.2f}% ({attrition_count} orang'
    ' resign)'
)
```

```python
#Bar Chart
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Department', palette='Blues_r')
plt.title('Jumlah Karyawan per Departemen')
plt.xlabel('Departemen')
plt.ylabel('Jumlah Karyawan')
plt.tight_layout()
plt.show()
```

```python
#Line Chart
tren_gaji_level = df.groupby('JobLevel')['MonthlyIncome'].mean().reset_index()

plt.figure(figsize=(7, 4))
plt.plot(tren_gaji_level['JobLevel'].astype(str), tren_gaji_level['MonthlyIncome'], marker='o', color='crimson')
plt.title('Tren Rata-rata Gaji Bulanan per Job Level')
plt.xlabel('Tingkat Jabatan (Job Level)')
plt.ylabel('Rata-rata Gaji Bulanan ($)')
plt.tight_layout()
plt.show()
```

```python
#Scatter Plot
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='Age', y='MonthlyIncome', hue='Attrition', alpha=0.7)
plt.title('Hubungan Usia dan Pendapatan Bulanan (Status Attrition)')
plt.xlabel('Usia (Tahun)')
plt.ylabel('Pendapatan Bulanan ($)')
plt.tight_layout()
plt.show()
```

```python
#Histogram
plt.figure(figsize=(7, 4))
sns.histplot(df['Age'], bins=15, kde=True, color='seagreen')
plt.title('Distribusi Frekuensi Usia Karyawan')
plt.xlabel('Usia (Tahun)')
plt.ylabel('Frekuensi')
plt.tight_layout()
plt.show()
```

```python
#Boxplot
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome', palette='Set2')
plt.title('Sebaran Gaji Bulanan Berdasarkan Status Resign')
plt.xlabel('Status Resign (Attrition)')
plt.ylabel('Gaji Bulanan ($)')
plt.tight_layout()
plt.show()
```

```python
#Dashboard Subplot 2x2
fig, axes = plt.subplots(2, 2, figsize=(14, 9))

# 1. Bar Chart (Kiri Atas)
sns.countplot(data=df, x='Department', palette='Blues_r', ax=axes[0, 0])
axes[0, 0].set_title('1. Jumlah Karyawan per Departemen')

# 2. Line Chart (Kanan Atas)
axes[0, 1].plot(tren_gaji_level['JobLevel'].astype(str), tren_gaji_level['MonthlyIncome'], marker='o', color='crimson')
axes[0, 1].set_title('2. Tren Rata-rata Gaji per Job Level')

# 3. Histogram (Kiri Bawah)
sns.histplot(df['Age'], bins=15, kde=True, color='seagreen', ax=axes[1, 0])
axes[1, 0].set_title('3. Distribusi Usia Karyawan')

# 4. Boxplot (Kanan Bawah)
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome', palette='Set2', ax=axes[1, 1])
axes[1, 1].set_title('4. Sebaran Gaji vs Attrition')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Menggunakan DataFrame 'df' yang sudah dimuat di sel sebelumnya
# Menampilkan nama-nama kolom yang tersedia
print('Daftar Kolom:', df.columns.tolist())

# Memilih kolom numerik untuk dideteksi outlier-nya (contoh: 'MonthlyIncome')
kolom_target = 'MonthlyIncome'

# 1. Visualisasi menggunakan Boxplot
plt.figure(figsize=(9, 5))
sns.boxplot(x=df[kolom_target], color='skyblue')
plt.title(f'Boxplot untuk Deteksi Outlier pada {kolom_target}', fontsize=14)
plt.xlabel(kolom_target, fontsize=12)
plt.show()

# 2. Perhitungan Batas IQR (Interquartile Range)
Q1 = df[kolom_target].quantile(0.25)
Q3 = df[kolom_target].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f'Q1 (Kuartil 1): {Q1}')
print(f'Q3 (Kuartil 3): {Q3}')
print(f'IQR: {IQR}')
print(f'Batas Bawah: {lower_bound}')
print(f'Batas Atas: {upper_bound}')

# 3. Menyaring dan Menampilkan Baris Data Outlier
outliers = df[
    (df[kolom_target] < lower_bound) | (df[kolom_target] > upper_bound)
]
print(f'Jumlah potensi outlier ditemukan: {len(outliers)}')
display(outliers[[kolom_target, 'JobRole', 'Department', 'TotalWorkingYears']])
```
